```python
library={}

def add():
    book_id=int(input("enter book id"))
    title=input("enter book name")
    author=input("enter book author")
    status=input("enter book status yes/no")
    library[book_id]={'title':title,'author':author,'status':status}

def show():
    print(library)
    
def check_status():
    book_id=int(input("enter book id"))
    if book_id in library:
        print("Available")
    else:
        print("Not Available")
        
def get_detail():
    book_id=int(input("enter book id"))
    print(library.get(book_id))

def update_status():
    book_id=int(input("enter book id"))
    new_status=input("enter new status")
    library[book_id]['status']=new_status

def clear_library():
    library.clear()
    
def delete_book():
    book_id=int(input("enter book id"))
    library.pop(book_id)
    
while True:
    print("\n1.add \n2.show \n3.check status \n4.get detail \n5.update status \n6.clear library \n7.delete book \n8.exit")
    ch=int(input("enter your choice"))
    if ch==1:
        add()
    elif ch==2:
        show()
    elif ch==3:
        check_status()
    elif ch==4:
        get_detail()
    elif ch==5:
        update_status()
    elif ch==6:
        clear_library()
    elif ch==7:
        delete_book()
    else:
        break
```
