```python
employee={}

def add():
    eid=int(input("enter empid"))
    name=input("enter emp name")
    dep=input("enter dep name")
    salary=int(input("enter emp salary"))
    employee[eid]={'name':name,'dept':dep,'salary':salary}

def show():
    print(employee)

def search():
    depart=input("enter dep name")
    for eid,value in employee.items():
        if value['dept']==depart:
            print(eid)
            print(value['name'])
            print(value['salary'])
            
def increment():
    depart=input("enter dep name")
    incr=int(input("enter increment"))
    for value in employee.values():
        if value['dept']==depart:
            value['salary']=value['salary']+incr 
            
def detail():
    for eid,value in employee.items():
            print(eid)
            print(value['name'])
            print(value['dept'])
            print(value['salary'])
    
        
while True:
    print("\n1.add \n2.show \n3.search \n4.increment \n5.detail \n6.exit")
    ch=int(input("enter your choice"))
    if ch==1:
        add()
    elif ch==2:
        show()
    elif ch==3:
        search()
    elif ch==4:
        increment()
    elif ch==5:
        detail()   
    else:
        break
```
