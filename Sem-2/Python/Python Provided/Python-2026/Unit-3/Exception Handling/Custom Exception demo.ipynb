```python
class InsufficientFundsException(Exception):
    """Exception raised when withdrawal amount exceeds balance."""
    pass

class InvalidAmountException(Exception):
    """Exception raised when deposit or withdrawal amount is negative."""
    pass

class InvalidAccountException(Exception):
    """Exception raised when account number is invalid."""
    pass

class BankAccount:
    def __init__(self, account_number, balance=0):
        if not str(account_number).isdigit() or len(str(account_number)) != 10:
            raise InvalidAccountException("Invalid account number! Must be a 10-digit numeric value.")
        
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        try:
            if amount <= 0:
                raise InvalidAmountException("Deposit amount must be greater than zero.")
            self.balance += amount
            print(f"Deposited: ${amount:.2f}. New Balance: ${self.balance:.2f}")
        except InvalidAmountException as e:
            print(f"Error: {e}")

    def withdraw(self, amount):
        try:
            if amount <= 0:
                raise InvalidAmountException("Withdrawal amount must be greater than zero.")
            if amount > self.balance:
                raise InsufficientFundsException("Insufficient funds in your account.")
            self.balance -= amount
            print(f"Withdrew: ${amount:.2f}. Remaining Balance: ${self.balance:.2f}")
        except (InvalidAmountException, InsufficientFundsException) as e:
            print(f"Error: {e}")

    def calculate_interest(self, principal, rate, time):
        try:
            if rate == 0:
                raise ZeroDivisionError("Interest rate cannot be zero.")
            interest = (principal * rate * time) / 100
            print(f"Interest: ${interest:.2f}")
        except ZeroDivisionError as e:
            print(f"Error: {e}")

# Example Usage
try:
    account = BankAccount("1234567890", 500)  # Valid account number
    account.deposit(200)  # Valid deposit
    account.withdraw(800)  # Insufficient funds
    account.withdraw(-50)  # Invalid amount
    account.calculate_interest(1000, 0, 2)  # Zero division error
    invalid_account = BankAccount("12345ABC")  # Invalid account number
except InvalidAccountException as e:
    print(f"Account Error: {e}")
```
