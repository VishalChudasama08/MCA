```python

class BankError(Exception):
    """Base class for Banking related exceptions"""
    pass


class InsufficientBalanceError(BankError):
    """Raised when withdrawal amount exceeds balance"""
    pass


class MinimumBalanceError(BankError):
    """Raised when balance goes below minimum required balance"""
    pass


class InvalidAmountError(BankError):
    """Raised when deposit or withdrawal amount is invalid"""
    pass


class AccountNotFoundError(BankError):
    """Raised when account number does not exist"""
    pass
#===================================================================================

class BankAccount:
    MIN_BALANCE = 1000

    def __init__(self, account_number, holder_name, balance):
        self.account_number = account_number
        self.holder_name = holder_name
        
        if balance < BankAccount.MIN_BALANCE:
            raise MinimumBalanceError("Initial balance must be at least",BankAccount.MIN_BALANCE)            
        
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise InvalidAmountError("Deposit amount must be positive")
        
        self.balance += amount
        print(amount," deposited successfully.")

    def withdraw(self, amount):
        if amount <= 0:
            raise InvalidAmountError("Withdrawal amount must be positive")

        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient balance")

        if self.balance - amount < BankAccount.MIN_BALANCE:
            raise MinimumBalanceError("Initial balance must be at least",BankAccount.MIN_BALANCE)

        self.balance -= amount
        print(amount," withdrawn successfully.")

    def show_balance(self):
        print("Current Balance:",self.balance)
#============================================================================
try:
    account = BankAccount(101, "Rahul", 5000)

    account.deposit(2000)
    account.withdraw(7000)   

except InvalidAmountError as e:
    print("Invalid Amount Error:", e)

except InsufficientBalanceError as e:
    print("Insufficient Balance Error:", e)

except MinimumBalanceError as e:
    print("Minimum Balance Error:", e)

except BankError as e:
    print("General Bank Error:", e)

else:
    print("Transaction completed successfully.")

finally:
    print("Thank you for banking with us.")
```
