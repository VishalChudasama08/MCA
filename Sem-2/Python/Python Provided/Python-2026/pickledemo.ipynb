```python
import pickle

student={}

def add():
    sid=int(input("enter id"))
    name=input("enter name")
    marks=int(input("enter marks"))
    student[sid]={"name":name,"marks":marks}
    f=open("stu.pkl","wb")
    pickle.dump(student,f)
    f.close()

def display():
    f=open("stu.pkl","rb")
    s=pickle.load(f)
    print(s)
    f.close()

def update():
    sid=int(input("enter id"))
    new=int(input("enter update marks"))
    student[sid]["marks"]=new
    f=open("stu.pkl","wb")
    pickle.dump(student,f)
    f.close()

def delete():
    sid=int(input("enter id"))
    student.pop(sid)
    f=open("stu.pkl","wb")
    pickle.dump(student,f)
    f.close()

while True:
    print("\n1.add \n2.display \n3.update \n4.delete")
    ch=int(input("enter your choice"))
    if ch==1:
        add()
    elif ch==2:
        display()
    elif ch==3:
        update()
    elif ch==4:
        delete()
    else:
        break
    
```
